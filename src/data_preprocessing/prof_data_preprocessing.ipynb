```python
def inverted_index_to_text(inverted_index):
    if not inverted_index:
        return ""

    # Find the highest word position
    max_position = max(
        pos
        for positions in inverted_index.values()
        for pos in positions
    )

    # Create empty list for words
    words = [None] * (max_position + 1)

    # Put each word back at its original position
    for word, positions in inverted_index.items():
        for pos in positions:
            words[pos] = word

    # Join words into normal text
    return " ".join(words)
```

```python
import json

def openalex_abstract_from_string(json_string):
    # JSON string → Python dictionary
    data = json.loads(json_string)

    # Get inverted index
    inverted_index = data.get("abstract_inverted_index")

    if not inverted_index:
        return None

    # Find maximum word position
    max_position = max(
        pos
        for positions in inverted_index.values()
        for pos in positions
    )

    # Reconstruct word order
    words = [None] * (max_position + 1)

    for word, positions in inverted_index.items():
        for pos in positions:
            words[pos] = word

    return " ".join(words)
```

```python
import json
import pandas as pd


def inverted_index_to_text(inverted_index):
    """Convert OpenAlex abstract_inverted_index to normal text."""
    
    if not inverted_index:
        return None

    max_position = max(
        pos
        for positions in inverted_index.values()
        for pos in positions
    )

    words = [None] * (max_position + 1)

    for word, positions in inverted_index.items():
        for pos in positions:
            words[pos] = word

    return " ".join(word for word in words if word is not None)


def process_openalex_row(row):
    """Extract all required information from one OpenAlex row."""

    raw_json = row["raw_json"]

    # If raw_json is a JSON string, convert it to dict
    if isinstance(raw_json, str):
        try:
            raw_json = json.loads(raw_json)
        except json.JSONDecodeError:
            raw_json = {}

    if raw_json.get("type") != "dataset":
        # -------------------------
        # Basic information
        # -------------------------
        name = row.get("name")
        expertise = row.get("expertise")
        department = row.get("department")

        # -------------------------
        # Title
        # -------------------------
        title = raw_json.get("title")

        # -------------------------
        # Topics
        # -------------------------
        topics_display_name = set()

        for topic in raw_json.get("topics", []):
            if topic.get("field"):
                topics_display_name.add(
                    topic["field"].get("display_name")
                )

            if topic.get("domain"):
                topics_display_name.add(
                    topic["domain"].get("display_name")
                )

            if topic.get("subfield"):
                topics_display_name.add(
                    topic["subfield"].get("display_name")
                )

            if topic.get("display_name"):
                topics_display_name.add(
                    topic["display_name"]
                )

        # Remove None
        topics_display_name.discard(None)

        # -------------------------
        # Concepts
        # -------------------------
        concepts_display_name = set()

        for concept in raw_json.get("concepts", []):
            if concept.get("display_name"):
                concepts_display_name.add(
                    concept["display_name"]
                )

        # -------------------------
        # Primary Topics
        # -------------------------
        # Your original code was using concepts here.
        # Keeping it exactly as you wrote it.
        primary_topics_display_name = set()

        for concept in raw_json.get("concepts", []):
            if concept.get("display_name"):
                primary_topics_display_name.add(
                    concept["display_name"]
                )

        # -------------------------
        # Abstract
        # -------------------------
        abstract_inverted_index = raw_json.get(
            "abstract_inverted_index"
        )

        abstract = inverted_index_to_text(
            abstract_inverted_index
        )

    # -------------------------
    # Return one row
    # -------------------------
        return {
            "name": name,
            "expertise": expertise,
            "department": department,
            "title": title,
            "topics_display_name": topics_display_name,
            "concepts_display_name": concepts_display_name,
            "primary_topics_display_name": primary_topics_display_name,
            "abstract": abstract
        }
    else:
        pass
```

```python
final_df = df.apply(process_openalex_row, axis=1, result_type="expand")
# print(final_df)
final_df.head()
```
